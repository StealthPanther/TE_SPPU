```python
import pandas as pd        # for dataframes and data handling
import numpy as np         # for numerical operations
import seaborn as sns      # for built-in datasets and plots
import matplotlib.pyplot as plt  # for plotting graphs
```

```python
# Fill missing 'age' values with median (middle value)
df['age'] = df['age'].fillna(df['age'].median())
```

```python
# Fill missing 'embarked' with most common value (mode)
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df['deck'] = df['deck'].cat.add_categories('Unknown')
df['deck'] = df['deck'].fillna('Unknown')
```

```python
# Convert 'age' from float to integer
df['age'] = df['age'].astype(int)

# Convert 'fare' from float to integer
df['fare'] = df['fare'].astype(int)
```

```python
# Verify no missing values remain
print("\n--- MISSING VALUES AFTER CLEANING ---")
print(df.isnull().sum())
```

```python
print("\n--- DATA TYPES AFTER CONVERSION ---")
print(df.dtypes)
```

```python
# --------------------------
# STEP 6: Turn Categorical Variables into Quantitative
# --------------------------

# Method 1: Using map() - manual label encoding
df['sex'] = df['sex'].map({'male': 1, 'female': 0})

df['embarked'] = df['embarked'].map({'S': 0, 'C': 1, 'Q': 2})

df['who'] = df['who'].map({'man': 0, 'woman': 1, 'child': 2})

df['alive'] = df['alive'].map({'yes': 1, 'no': 0})
```
