```python
   # =====================================================
# ASSIGNMENT 10 - Data Analytics III
# Naive Bayes Classification on Iris Dataset
# =====================================================

# STEP 1: Imports (only needed!)
import seaborn as sns
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import multilabel_confusion_matrix
from sklearn.preprocessing import LabelEncoder      
```

```python
# STEP 2: Load Dataset (built-in!)
df = sns.load_dataset('iris')
print("Dataset Loaded!")
print(df.head())
```

```python
# STEP 3: X and Y
X = df.drop('species', axis=1)   # 4 features
Y = df['species']                 # species name

# Convert species text to numbers (setosa=0, versicolor=1, virginica=2)
le = LabelEncoder()
Y = le.fit_transform(Y)
print(le.classes_)
```

```python
# STEP 4: Split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
    # STEP 5: Train Naive Bayes Model
model = GaussianNB()
model.fit(X_train, Y_train)
print("\nModel Trained!")

# STEP 6: Predict
Y_pred = model.predict(X_test)
print("Predictions:", Y_pred)
print("Actual     :", Y_test)
```

```python
mcm = multilabel_confusion_matrix(Y_test, Y_pred)
print("\nmultilabel Confusion Matrix:")
print(mcm)
```

```python

# Ek line mein saare classes ka TP/TN/FP/FN!
mcm = multilabel_confusion_matrix(Y_test, Y_pred)

# classes = ['setosa', 'versicolor', 'virginica']

for i in range(3):
    TN = mcm[i][0][0]
    FP = mcm[i][0][1]
    FN = mcm[i][1][0]
    TP = mcm[i][1][1]

    accuracy  = (TP + TN) / (TP + TN + FP + FN)
    error     = 1 - accuracy
    precision = TP / (TP + FP)
    recall    = TP / (TP + FN)

    print("\nClass:", le.classes_[i])
    print("  TP:", TP, " TN:", TN, " FP:", FP, " FN:", FN)
    print("  Accuracy :", round(accuracy, 2))
    print("  Error    :", round(error, 2))
    print("  Precision:", round(precision, 2))
    print("  Recall   :", round(recall, 2))
```
