```python
# =====================================================
# ASSIGNMENT 7 - Text Analytics
# PART 1: Text Preprocessing
# =====================================================

import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
```

```python
# Sample Document
text = "The students are studying hard for their exams. They are learning Python programming and data science. Good students always practice regularly."
```

```python
print("Original Text:")
print(text)
```

```python
# -----------------------------------------------
# STEP 1: Tokenization
# -----------------------------------------------
# Word Tokenization - text ko words mein todo
word_tokens = word_tokenize(text)
print("\n--- Word Tokens ---")
print(word_tokens)
```

```python
# Sentence Tokenization - text ko sentences mein todo
sent_tokens = sent_tokenize(text)
print("\n--- Sentence Tokens ---")
print(sent_tokens)
```

```python
# -----------------------------------------------
# STEP 2: POS Tagging
# -----------------------------------------------
# Har word ko grammar tag do (Noun, Verb, etc.)
pos_tags = pos_tag(word_tokens)
print("\n--- POS Tags ---")
print(pos_tags)
```

```python
# -----------------------------------------------
# STEP 3: Stop Words Removal
# -----------------------------------------------
# Common words hata do (the, is, are, a, an...)
stop_words = set(stopwords.words('english'))

filtered = []
for word in word_tokens:
    if word.lower() not in stop_words:
        filtered.append(word)

print("After Stop Words Removal:")
print(filtered)
```

```python
# STEP 4: Stemming
stemmer = PorterStemmer()

stemmed = []
for word in filtered:
    stemmed.append(stemmer.stem(word))

print("After Stemming:")
print(stemmed)
```

```python
# STEP 5: Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized = []
for word in filtered:
    lemmatized.append(lemmatizer.lemmatize(word))

print("After Lemmatization:")
print(lemmatized)
```

```python
# =====================================================
# PART 2: TF-IDF using sklearn (simplest way!)
# =====================================================

from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
# Our documents (3 simple sentences)
documents = [
    "students study Python programming",
    "data science is interesting",
    "students learn data science and Python"
]
```

```python
# Create TF-IDF
tfidf = TfidfVectorizer()
result = tfidf.fit_transform(documents)
```

```python
# Show as a readable table
df_tfidf = pd.DataFrame(
    result.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("TF-IDF Matrix:")
print(df_tfidf)
```
