```python
# =====================================================
# ASSIGNMENT 6 - Data Visualization III
# Dataset: Iris Flower Dataset
# Source: https://archive.ics.uci.edu/ml/datasets/Iris
# =====================================================

# STEP 1: Import Libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# STEP 2: Load Dataset (built-in in seaborn)
df = sns.load_dataset('iris')
print("Dataset Loaded!")
print(df.head())
```

```python
print("\nShape:", df.shape)
```

```python
# =====================================================
# STEP 3: Features and Their Types
# =====================================================
print("\n--- Feature Types ---")
print(df.dtypes)
```

```python
# =====================================================
# STEP 4: Histogram for Each Feature
# =====================================================

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for feature in features:
    sns.histplot(df[feature], bins=10, kde=True, color='blue')
    plt.title("Histogram of " + feature)
    plt.xlabel(feature)
    plt.ylabel("Count")
    plt.show()
```

```python
# =====================================================
# STEP 5: Box Plot for Each Feature
# =====================================================

for feature in features:
    sns.boxplot(y=df[feature], color='green')
    plt.title("Box Plot of " + feature)
    plt.ylabel(feature)
    plt.show()
```

```python
# BONUS: All box plots together (easy comparison)
df.boxplot()
plt.title("Box Plot - All Features")
plt.show()
```
