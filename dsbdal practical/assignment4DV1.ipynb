```python
# =====================================================
# ASSIGNMENT 4 - Data Visualization I
# Dataset: Titanic (built-in seaborn dataset)
# =====================================================

# STEP 1: Import Libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# STEP 2: Load Dataset
df = sns.load_dataset('titanic')
print("Dataset Loaded! Shape:", df.shape)
print(df.head())
```

```python
# =====================================================
# PART 1 - Finding Patterns using Seaborn Plots
# =====================================================

# PLOT 1: Count of Survivors (countplot)
# Pattern: How many people survived vs not survived?
sns.countplot(x='survived', data=df)
plt.title("Survival Count (0=No, 1=Yes)")
plt.show()
```

```python
# PLOT 2: Survival by Gender
# Pattern: Did gender affect survival?
sns.countplot(x='survived', hue='sex', data=df)
plt.title("Survival by Gender")
plt.show()
```

```python
# PLOT 3: Survival by Passenger Class
# Pattern: Did ticket class affect survival?
sns.countplot(x='pclass', hue='survived', data=df)
plt.title("Survival by Passenger Class")
plt.show()
```

```python
# PLOT 4: Age Distribution
# Pattern: What was the age distribution of passengers?
sns.histplot(df['age'].dropna(), bins=10, color='blue')
plt.title("Age Distribution of Passengers")
plt.show()
```

```python
# PLOT 5: Fare vs Age (Scatter Plot)
# Pattern: Did older passengers pay more?
sns.scatterplot(x='age', y='fare', hue='sex', data=df)
plt.title("Fare vs Age")
plt.show()
```

```python
# PLOT 6: Heatmap - Correlation between numeric columns
# Pattern: Which columns are related to each other?
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# =====================================================
# PART 2 - Histogram of Fare Column
# =====================================================
sns.histplot(df['fare'], bins=20, color='green', kde=True)
plt.title("Distribution of Ticket Fare")
plt.xlabel("Fare Price")
plt.ylabel("Number of Passengers")
plt.show()
```
