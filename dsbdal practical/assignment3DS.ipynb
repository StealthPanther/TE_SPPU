```python
# =====================================================
# ASSIGNMENT 3 - Descriptive Statistics
# Part 1: Summary stats grouped by Grade (categorical)
# =====================================================

# STEP 1: Import Libraries
import pandas as pd
import numpy as np

# STEP 2: Create Dataset
data = {
    'Name':       ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
                   'Frank', 'Grace', 'Henry', 'Isla', 'Jake'],
    'Age':        [20, 21, 22, 22, 20, 23, 21, 21, 22, 20],
    'Marks':      [85, 90, 78, 88, 92, 76, 95, 82, 80, 91],
    'Attendance': [90, 85, 78, 88, 92, 78, 95, 80, 85, 88],
    'StudyHours': [5, 6, 4, 7, 5, 3, 8, 4, 6, 7],
    'Grade':      ['A', 'A', 'B', 'A', 'A', 'B', 'A', 'B', 'B', 'A']
}
df = pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
# STEP 3: Summary Statistics (mean, median, min, max, std)
# Grouped by 'Grade' (categorical variable)
# Numeric variable: Marks
# =====================================================

print("\n--- Summary Statistics of Marks grouped by Grade ---")
grouped = df.groupby('Grade')['Marks']

print("\nMean:")
print(grouped.mean())

print("\nMedian:")
print(grouped.median())

print("\nMinimum:")
print(grouped.min())

print("\nMaximum:")
print(grouped.max())

print("\nStandard Deviation:")
print(grouped.std())
```

```python
# All at once using describe()
print("\n--- Full Summary using describe() grouped by Grade ---")
print(df.groupby('Grade')['Marks'].describe())
```

```python
# =====================================================
# STEP 4: List of numeric values for each Grade category
# =====================================================
print("\n--- List of Marks for each Grade ---")
for grade, group in df.groupby('Grade'):
    print("Grade", grade, ":", list(group['Marks']))
```

```python
# =====================================================
# PART 2: Iris Dataset - Statistics by Species
# =====================================================

# Load Iris dataset (built-in from seaborn, no download needed!)
import seaborn as sns

iris = sns.load_dataset('iris')
print("\nIris Dataset - First 5 rows:")
print(iris.head())

print("\nColumn Names:")
print(iris.columns.tolist())
```

```python
print(iris.columns)
```

```python
# =====================================================
# Basic Stats for each species
# =====================================================
print("\n--- Statistics grouped by Species ---")
print(iris.groupby('species').describe())
```

```python
# Individual stats for Iris-setosa
print("\n--- Stats for setosa only ---")
setosa = iris[iris['species'] == 'setosa']
print("Mean:\n", setosa.mean(numeric_only=True),"\n")
print("Median:\n", setosa.median(numeric_only=True),"\n")
print("Std Dev:\n", setosa.std(numeric_only=True),"\n")
print("Min:\n", setosa.min(numeric_only=True),"\n")
print("Max:\n", setosa.max(numeric_only=True),"\n")
```

```python
# Percentile
print("\n--- Percentile (25th, 50th, 75th) for setosa ---")
print(setosa.quantile([0.25, 0.50, 0.75], numeric_only=True))
```

```python
# All 3 species at once
print("\n--- Full describe() for all species ---")
print(iris.groupby('species').describe())
```
